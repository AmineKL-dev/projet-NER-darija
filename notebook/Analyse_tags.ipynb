```python
import re
import pandas as pd
import os

BASE_DIR = os.getcwd()

# 2. On construit le chemin vers 'data' de manière robuste
# Cela fonctionnera même sous Windows ou Linux (gestion des slashs / vs \)
data_path = os.path.join(BASE_DIR, "../data")

def extract_ner_contexts(directory_path, target_tag=None, target_value=None):
    # Pattern pour trouver les tags et leur contenu
    pattern = r"<(\w+)>\s*(.*?)\s*</\1>"
     # Vérifier si le dossier existe
    if not os.path.exists(directory_path):
        print(f"Erreur : Le dossier '{directory_path}' est introuvable.")
        return None
    
    for filename in os.listdir(directory_path):
        if filename.endswith(".txt"):
            file_path = os.path.join(directory_path, filename)
            # On découpe par ligne pour isoler les phrases
            with open(file_path, 'r', encoding='utf-8') as f:
                file_content = f.read()
            lines = file_content.split('\n')
            results = []

            for line in lines:
                if not line.strip(): continue
                
                matches = re.finditer(pattern, line)
                for match in matches:
                    tag = match.group(1)
                    value = match.group(2).strip()
                    
                    # Filtres optionnels (ex: focus sur 'ITEM' ou 'dollar')
                    if target_tag and tag != target_tag: continue
                    if target_value and target_value.lower() not in value.lower(): continue
                    
                    results.append({
                        "Tag": tag,
                        "Valeur": value,
                        "Phrase_Complete": line.strip()
                    })
            
    return pd.DataFrame(results)

# Exemple pour analyser le problème "dollar"
# df_check = extract_ner_contexts(votre_texte, target_value="dollar")
```
